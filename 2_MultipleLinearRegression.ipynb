Given the following data, apply a Multiple Linear Regression model to predict \(Y\) using two independent variables \(X_1\) and \(X_2\).

Tasks:

Construct the input matrix \(X\) and output vector \(Y\).

Calculate the regression coefficients using:

$$ a=(X^TX)^{-1}X^TY $$

Obtain the regression equation:

$$ \hat Y=a_0+a_1X_1+a_2X_2 $$
Use the obtained model to predict \(Y\) for a new pair of values \(X_1\) and \(X_2\).

Data:

$$ X_1=[1,2,3,4] $$ $$ X_2=[9,5,8,2] $$
Y=[1,6,8,12]

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X= np.array([[1,4],[2,5],[3,8],[4,2]])
Y = np.array([1,6,8,12])
```

```python
model= LinearRegression()
model.fit(X,Y)
```

```python
#Get Coefficient
a0=model.intercept_
a1=model.coef_[0]
a2=model.coef_[1]

print("a0 =", a0)
print("a1 =", a1)
print("a2 =", a2)
```

```python
#Regression Equation
print("\nMultiple Linear Regression Equation:")
print(f"Y={a0:.3f}+{a1:.3f}X1+{a2:.3f}X2")
```

```python
Y_pred= model.predict(X)
print("\nActual Y:")
print(Y)
print("\nPredict Y:")
print(Y_pred)
```

```python
# Example:
# X1 = 5
# X2 = 6
new_data = np.array([[5,6]])
prediction = model.predict(new_data)
print("\nPrediction for new X1, and X2")
print("predicted Y =", prediction[0])
```

```python
#Plot the data
fig = plt.figure(figsize=(9,7))
ax=fig.add_subplot(111,projection='3d')
ax.scatter(X[:,0],X[:,1],Y,s=80,label="Acctual data")

#Create Reggression Plane
X1_range = np.linspace(X[:,0].min(), X[:,0].max(), 20)
x2_range = np.linspace(X[:,1].min(), X[:,1].max(), 20)
X1,X2 = np.meshgrid(X1_range, x2_range)
Y_plane=(a0 + a1* X1 + a2*X2)

#plot reggression plane
ax.plot_surface(X1,X2,Y_plane, alpha=0.5)
ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("Y")

ax.set_title("Multiple Linear Regression")

ax.legend()

plt.show()
```
