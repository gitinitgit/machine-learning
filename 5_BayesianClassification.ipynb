### Problem Statement — Bayesian Classifier

Consider the following dataset of **14 customers**. The objective is to predict whether a customer has the **ability to buy** a product using a **Naive Bayes/Bayesian Classifier**.

| No. | Age   | Income | Student | Credit Score | Buying Ability |
| --: | ----- | ------ | ------- | ------------ | -------------- |
|   1 | ≤30   | High   | No      | Fair         | No             |
|   2 | ≤30   | High   | No      | Excellent    | No             |
|   3 | 31–40 | High   | No      | Fair         | Yes            |
|   4 | >40   | Medium | No      | Fair         | Yes            |
|   5 | >40   | Low    | Yes     | Fair         | Yes            |
|   6 | >40   | Low    | Yes     | Excellent    | No             |
|   7 | 31–40 | Low    | Yes     | Excellent    | Yes            |
|   8 | ≤30   | Medium | No      | Fair         | No             |
|   9 | ≤30   | Low    | Yes     | Fair         | Yes            |
|  10 | >40   | Medium | Yes     | Fair         | Yes            |
|  11 | ≤30   | Medium | Yes     | Excellent    | Yes            |
|  12 | 31–40 | Medium | No      | Excellent    | Yes            |
|  13 | 31–40 | High   | Yes     | Fair         | Yes            |
|  14 | >40   | Medium | No      | Excellent    | No             |

A new customer has the following attributes:

$$
X = [\text{Age}\leq30,\; Income=Medium,\; Student=Yes,\; Credit=Fair]
$$

Using the **Naive Bayes classifier**, determine whether this customer belongs to:

* \(C_1\): **Buying Ability = Yes**
* \(C_2\): **Buying Ability = No**

Calculate:

$$
P(C_1|X) \quad \text{and} \quad P(C_2|X)
$$

using Bayes' theorem:

$$
P(C_i|X)=\frac{P(X|C_i)P(C_i)}{P(X)}
$$

and classify the customer into the class having the **higher posterior probability**.

### From the given dataset

There are **9 Yes** and **5 No**:

$$
P(C_1)=\frac{9}{14},\qquad
P(C_2)=\frac{5}{14}
$$

For the new customer, calculate:

$$
P(X|C_1)
=
P(Age\leq30|C_1)
P(Income=Medium|C_1)
P(Student=Yes|C_1)
P(Credit=Fair|C_1)
$$

and similarly for \(C_2\).

This is the exact problem your notes are solving.


```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB
```

```python
# Data set
data = {
    "Age": [
        "<=30", "<=30", "31-40", ">40", ">40", ">40",
        "31-40", "<=30", "<=30", ">40", "<=30", "31-40",
        "31-40", ">40"
    ],

    "Income": [
        "High", "High", "High", "Medium", "Low", "Low",
        "Low", "Medium", "Low", "Medium", "Medium", "Medium",
        "High", "Medium"
    ],

    "Student": [

        "No", "No", "No", "No", "Yes", "Yes",
        "Yes", "No", "Yes", "Yes", "Yes", "No",
        "Yes", "No"
    ],

    "Credit": [
        "Fair", "Excellent", "Fair", "Fair", "Fair", "Excellent",
        "Excellent", "Fair", "Fair", "Fair", "Excellent", "Excellent",
        "Fair", "Excellent"
    ],

    "Buying": [
        "No", "No", "Yes", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "Yes", "Yes",
        "Yes", "No"
    ]
}

df = pd.DataFrame(data)
print("Dataset:")
display(df)
```

```python
# 2. Separate features & target

X = df[["Age","Income","Student", "Credit"]]
y = df["Buying"]

# 3. Encode categorical data

encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)
print("\nEncoded X: ")
print(X_encoded)
```

```python
# 4. Train Naive Bayes

model = CategoricalNB(alpha=1e-10)
model.fit(X_encoded, y)
```

```python
# 5. New customer
new_customer=pd.DataFrame({
    "Age": ["<=30"],
    "Income": ["Medium"],
    "Student": ["Yes"],
    "Credit": ["Fair"]
})
new_customer_encoded = encoder.transform(new_customer)
```

```python
# 6. Prediction
prediction = model.predict(new_customer_encoded)
print("\nPrediction")
print("Buying Ability =", prediction[0])
```

```python
# 7. Probabilities
probabilities = model.predict_proba(new_customer_encoded)
print("\nProbablities:")
for class_name, probability in zip(model.classes_, probabilities[0]):
    print(f"P({class_name} | X) = {probability:.4f}")



```

```python
#8. Final Result

print("\nFinal Answer:")

if prediction[0] == "Yes":
    print("The customer has Buying Ability = YES")
else:
    print("The customer has Buying Ability = NO")
```
