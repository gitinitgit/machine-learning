1. Problem Statement

Consider a binary classification problem with two training samples:

$$ X_1= \begin{bmatrix} 2\\ 1 \end{bmatrix}, \qquad y_1=+1 $$

and

$$ X_2= \begin{bmatrix} 4\\ 3 \end{bmatrix}, \qquad y_2=-1 $$

We want to construct a linear Support Vector Machine that separates the two classes with a maximum margin.

The objective is to find:

$$ \boxed{f(x)=w^Tx+b} $$

where:

\(w\) = weight vector
\(b\) = bias/intercept
\(y=+1\) and \(y=-1\) are the two classes.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

# -----------------------------
# 1. Dataset
# -----------------------------

X = np.array([
    [2, 1],
    [4, 3]
])

y = np.array([
    1,
    -1
])

# -----------------------------
# 2. Create Linear SVM
# -----------------------------

svm = SVC(
    kernel='linear',
    C=1000
)

# -----------------------------
# 3. Train
# -----------------------------

svm.fit(X, y)

# -----------------------------
# 4. Get weights and bias
# -----------------------------

w = svm.coef_[0]
b = svm.intercept_[0]

print("Weight vector w =", w)
print("Bias b =", b)

# -----------------------------
# 5. Predictions
# -----------------------------

predictions = svm.predict(X)

print("\nActual values:    ", y)
print("Predicted values: ", predictions)

# -----------------------------
# 6. Decision function
# -----------------------------

decision_values = svm.decision_function(X)

print("\nDecision values:", decision_values)

# -----------------------------
# 7. Support vectors
# -----------------------------

print("\nSupport vectors:")
print(svm.support_vectors_)

# -----------------------------
# 8. Accuracy
# -----------------------------

accuracy = svm.score(X, y)

print("\nAccuracy:", accuracy)
```

plot the Decision Boundary

```python
plt.figure(figsize=(7, 5))

# Plot positive and negative samples
plt.scatter(
    X[y == 1, 0],
    X[y == 1, 1],
    marker='o',
    s=100,
    label='Class +1'
)

plt.scatter(
    X[y == -1, 0],
    X[y == -1, 1],
    marker='s',
    s=100,
    label='Class -1'
)

# Decision boundary:
# w1*x1 + w2*x2 + b = 0
x_values = np.linspace(0, 6, 100)

y_boundary = -(w[0] * x_values + b) / w[1]

plt.plot(
    x_values,
    y_boundary,
    label='Decision Boundary'
)

plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Linear SVM")
plt.legend()
plt.grid()

plt.show()
```
